# MVS Python quick start
This notebook uses the Python package as a client of the same MVS .NET engine. The example data are synthetic/demo data; successful execution is not scientific validation.


```python
# Development install from the repository root:
# %pip install -e ../python
import mvs
engine = mvs.configure_engine()  # or pass an explicit path to the mvs executable
engine.identity.raw['engineVersion']
```

```python
calibration = mvs.calibrate(
    '../examples/demo_three_groups.csv',
    repetitions=150,
    seed=20260719,
)
calibration.output_directory
```

```python
result = mvs.analyze(
    '../examples/demo_three_groups.csv',
    calibration=calibration,
)
result.read_csv('results.csv')[:3]
```

The saved `run_manifest.json` and result artifacts remain the authoritative reproducibility record. Inspect warnings, diagnostics, methods and validation limitations before interpreting results.
